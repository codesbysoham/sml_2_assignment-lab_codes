# GA Question 1 — Maximize \(f(n)=n^2\)

5-bit binary encoding gives \(0\le n\le31\).

Initial population: `11011, 10001, 01111, 10111`

Use \(P_i=f_i/\sum f_i\), \(E_i=f_i/\bar f\).

```python
import math
import pandas as pd

population = ["11011", "10001", "01111", "10111"]

def decode(bits):
    return int(bits, 2)

def fitness(bits):
    return decode(bits)**2

n_values = [decode(s) for s in population]
fit = [fitness(s) for s in population]
total = sum(fit)
average = total / len(population)

table = pd.DataFrame({
    "Chromosome": population,
    "n": n_values,
    "Fitness f(n)=n^2": fit
})
table["Selection Probability"] = table["Fitness f(n)=n^2"] / total
table["Expected Count"] = table["Fitness f(n)=n^2"] / average
table.round(4)
```

## Solution

Decoded values: \(11011_2=27\), \(10001_2=17\), \(01111_2=15\), \(10111_2=23\).

Fitness: \(729,289,225,529\).

Total fitness: \(1772\). Average fitness: \(443\).

Expected counts: \(1.646,0.652,0.508,1.194\).

**Note:** \(289/443=0.6524\), so an expected count of 0.638 would be incorrect.

The global maximum over \(0\le n\le31\) is \(n=31\), represented by `11111`, with \(f(31)=961\).

```python
import random

random.seed(42)

def crossover(p1, p2, point):
    return p1[:point] + p2[point:], p2[:point] + p1[point:]

def mutate(bits, rate=0.10):
    bits = list(bits)
    for i in range(len(bits)):
        if random.random() < rate:
            bits[i] = '1' if bits[i] == '0' else '0'
    return ''.join(bits)

mating_pool = ["11011", "11011", "10111", "10111"]
offspring = []
for i in range(0, 4, 2):
    c1, c2 = crossover(mating_pool[i], mating_pool[i+1], 3)
    offspring.extend([c1, c2])

new_population = [mutate(s) for s in offspring]
print("Mating pool:", mating_pool)
print("Offspring:", offspring)
print("After mutation:", new_population)
best = max(new_population, key=fitness)
print("Best chromosome:", best)
print("Best n:", decode(best))
print("Best fitness:", fitness(best))
```
