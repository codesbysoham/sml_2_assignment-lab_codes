# GA Question 2 — Maximize \(f(n)=-n^2+2n\) with real-valued encoding

**Given:** \(0\le n\le2\), 5-bit chromosomes `11010, 00111, 10110, 00101`, random numbers `0.4, 0.15, 0.7, 0.9`, and crossover at the **5th digit**.

Real-valued decoding: \(n=2D/31\), where \(D\) is the decimal value of the chromosome.

```python
import pandas as pd
import numpy as np

population = ["11010", "00111", "10110", "00101"]

def decode(bits):
    D = int(bits, 2)
    return 2 * D / 31

def fitness(n):
    return -n**2 + 2*n

decoded = [decode(s) for s in population]
fit = [fitness(n) for n in decoded]
total = sum(fit)

table = pd.DataFrame({
    "Chromosome": population,
    "Decimal D": [int(s, 2) for s in population],
    "n": decoded,
    "f(n)": fit
})
table["Probability"] = table["f(n)"] / total
table["Cumulative Probability"] = table["Probability"].cumsum()

print(table.round(4))
print("\nTotal fitness:", round(total, 4))
```

## Solution

Decoded values: approximately \(1.6774,0.4516,1.4194,0.3226\).

Fitness values: approximately \(0.5411,0.6993,0.8241,0.5411\).

Total fitness: \(2.6056\). Selection probabilities: approximately \(0.2077,0.2684,0.3163,0.2077\).

Cumulative probabilities: \(0.2077,0.4760,0.7923,1.0000\).

Given random numbers select: \(0.40\to00111\), \(0.15\to11010\), \(0.70\to10110\), \(0.90\to00101\).

Crossover at the 5th digit produces no change because the chromosome has only 5 bits.

Analytically, \(f(n)=-(n-1)^2+1\), so the continuous optimum is \(\boxed{n=1,f(n)=1}\). The 5-bit mapping does not represent exactly \(n=1\); the nearest encoded values are approximately 0.9677 and 1.0323.

```python
random_numbers = [0.40, 0.15, 0.70, 0.90]
cumulative = table["Cumulative Probability"].tolist()

selected = []
for r in random_numbers:
    for i, c in enumerate(cumulative):
        if r <= c:
            selected.append(population[i])
            break

print("Selected population:", selected)

def crossover(p1, p2, point=5):
    return p1[:point] + p2[point:], p2[:point] + p1[point:]

children = []
for i in range(0, len(selected), 2):
    c1, c2 = crossover(selected[i], selected[i+1], 5)
    children.extend([c1, c2])

print("Children after crossover at digit 5:", children)
```
