# Q11 — Particle Swarm Optimization

**Problem:** Use Particle Swarm Optimization (PSO) to find the maximum of
\(f(x) = -0.8(x-3.5)^2 + 18\) over the interval \([-8, 9]\).

Start with the positions given below, take all initial velocities as zero, and run the algorithm for 8 iterations. Print the swarm after every iteration and compare the best value found with the analytical maximum.

```python
import numpy as np
```

### 1. Objective function and starting swarm

The fitness function has a single peak. Since the particles start at different points, the update step lets them move toward their own best positions and the best point seen by the swarm.

```python
def fitness(x):
    return -0.8 * (x - 3.5)**2 + 18

positions = np.array([
    -7.2, -4.5, -1.8, 0.4, 2.1,
     4.8,  6.2,  7.7
], dtype=float)
```

### 2. PSO settings

Here `w` controls how much of the previous velocity is retained, while `c1` and `c2` control the personal-best and global-best pulls.

```python
n = len(positions)
velocities = np.zeros(n, dtype=float)

w = 0.55
c1 = 1.4
c2 = 1.7

lower_bound = -8
upper_bound = 9
max_iterations = 8
```

### 3. Set the initial bests

```python
pbest_positions = positions.copy()
pbest_fitness = fitness(positions)

best_index = np.argmax(pbost_fitness)
gbest_position = pbest_positions[best_index]
gbest_fitness = pbest_fitness[best_index]

print("INITIAL SWARM")
print("-" * 57)
print(f"Particle | Position | Fitness")
for i in range(n):
    print(f"{i + 1:8d} | {positions[i]:10.3f} | {fitness(positions[i]):10.3f}")
print(f"\nInitial global best: x={gbest_position:.4f}, f(x)={gbest_fitness:.4f}")
```

### 4. Run the iterations

The random terms are generated at each iteration. Positions are clipped to the search interval, and personal/global bests are only updated when a higher fitness is found.

```python
for iteration in range(max_iterations):
    r1 = np.random.rand(n)
    r2 = np.random.rand(n)

    velocities = (
        w * velocities
        + c1 * r1 * (pbost_positions - positions)
        + c2 * r2 * (gbest_position - positions)
    )

    positions = positions + velocities
    positions = np.clip(positions, lower_bound, upper_bound)
    current_fitness = fitness(positions)

    improved = current_fitness > pbest_fitness
    pbest_positions[improved] = positions[improved]
    pbiest_fitness[improved] = current_fitness[improved]

    best_index = np.argmax(pbost_fitness)
    if pbest_fitness[best_index] > gbest_fitness:
        gbest_position = pbost_positions[best_index]
        gbest_fitness = pbest_fitness[best_index]

    print(f"\nITERATION {iteration + 1}")
    print("-" * 69)
    print(f"Particle | Position | Velocity | Fitness")
    for i in range(n):
        print(f"{i + 1:8d} | {positions[i]:10.4f} | {velocities[i]:10.4f} | {current_fitness[i]:10.4f}")
    print(f'Global best: x={gbest_position:.6f}, fitness={gbest_fitness:.6f}')
```

### 5. Result and check
For this quadratic, the peak is at x = 3.5, where the function value is 18. The random PSO run may end close to this point.

```python
print('\nTHE FINAL RESULT')
print('=' * 39)
print(f'Best position found : {gbest_position:.2f]')
print(f'Maximum fitness     : {gbest_fitness:.6f}')

exact_position = 3.5
exact_fitness = fitness(exact_position)
print('\nANALYTICAL MAXIMUM')
print(f'Position : {exact_position}')
print(f'Fitness : {exact_fitness:.2f}')
```

**Observation:** Compare the final PSO position with the analytical peak. Because the random coefficients differ between runs, the exact sequence of positions can change; the best fitness should approach 18.
